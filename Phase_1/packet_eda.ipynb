
# Packet-Level Exploratory Data Analysis

This notebook explores `packet_sample_seed1.csv` before preprocessing and model training.

The main goals are to:

- verify the sampled class distribution;
- inspect data types, missing values, duplicates, and constant columns;
- understand important packet, protocol, timing, and rolling-window features;
- identify skewed variables, outliers, and highly correlated features;
- record preprocessing decisions for the Phase 2 packet-level models.

`binary_label` and `attack_type` are used only for EDA and later evaluation. They must not be passed into unsupervised model training.


## Setup, load metadata and sampled packet dataset

```python

import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 60)
pd.set_option("display.max_rows", 100)

METADATA_FILE_NAME = "packet_sample_seed1.metadata.json"

PROJECT_ROOT = Path("..")
META_PATH = PROJECT_ROOT / "data" / "samples" / METADATA_FILE_NAME

with open(META_PATH, "r") as f:
    metadata = json.load(f)

DATA_PATH = PROJECT_ROOT / metadata["output_path"]
df = pd.read_csv(DATA_PATH, low_memory=False)

BINARY_COL = "binary_label"
ATTACK_COL = "attack_type"
SOURCE_COL = "source_file"

print("Dataset path:", DATA_PATH)
print("Dataset shape:", df.shape)
print("Memory usage: {:.2f} MB".format(df.memory_usage(deep=True).sum() / 1024**2))
df.head()
```


## Dataset overview

Start by checking whether the number of rows, number of columns, labels, and source files match the sampling metadata.


```python

overview = pd.DataFrame({
    "value": [
        len(df),
        df.shape[1],
        df[BINARY_COL].nunique(dropna=False),
        df[ATTACK_COL].nunique(dropna=False),
        df[SOURCE_COL].nunique(dropna=False)
    ]
}, index=[
    "Rows",
    "Columns",
    "Binary classes",
    "Traffic types",
    "Source files"
])

display(overview)
print("Metadata seed:", metadata.get("seed"))
print("Metadata total rows:", metadata.get("total_rows"))
print("Rows match metadata:", len(df) == metadata.get("total_rows"))
```

```python

binary_summary = df[BINARY_COL].value_counts().rename_axis(BINARY_COL).to_frame("count")
binary_summary["percentage"] = binary_summary["count"] / len(df) * 100

type_order = ["Benign", "DDoS-HTTP Flood", "DoS-HTTP Flood", "DNS Spoofing", "Brute Force", "XSS"]
attack_summary = df[ATTACK_COL].value_counts().reindex(type_order).dropna().rename_axis(ATTACK_COL).to_frame("count")
attack_summary["percentage"] = attack_summary["count"] / len(df) * 100

display(binary_summary.round(3))
display(attack_summary.round(3))

ax = attack_summary["count"].sort_values().plot(kind="barh", figsize=(9, 5), logx=True)

for i, value in enumerate(attack_summary["count"].sort_values()):
    ax.text(value * 1.05, i, f"{int(value):,}", va="center")

ax.set_xlabel("Packet count (log scale)")
ax.set_ylabel("Traffic type")
ax.set_title("Packet Distribution by Traffic Type")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```


The sampled dataset is intentionally imbalanced because benign traffic should represent roughly 97–98% of the packets. Accuracy alone will therefore be misleading. Later models should also report precision, recall, F1, PR-AUC, ROC-AUC, FPR, and FNR.

## Histograms of all candidate features

This section gives a compact visual overview of every candidate feature. Numeric features are shown as benign-versus-attack histograms. Categorical features are shown with the class composition of their most common values because regular histograms are not suitable for text categories.

The target columns `binary_label` and `attack_type` and the leakage column `source_file` are excluded. The plots use a fixed sample for speed. Numeric x-axes are limited to the 1st–99th percentile so extreme network values do not hide the main distribution.

```python
HIST_SAMPLE_PER_CLASS = 15000
NUMERIC_PLOTS_PER_PAGE = 12
CATEGORICAL_PLOTS_PER_PAGE = 4
TOP_CATEGORIES = 8

candidate_cols = [col for col in df.columns if col not in [BINARY_COL, ATTACK_COL, SOURCE_COL]]
numeric_feature_cols = [col for col in candidate_cols if pd.api.types.is_numeric_dtype(df[col])]
categorical_feature_cols = [col for col in candidate_cols if col not in numeric_feature_cols]

hist_parts = []
for label, group in df.groupby(BINARY_COL):
    sample_size = min(len(group), HIST_SAMPLE_PER_CLASS)
    hist_parts.append(group.sample(sample_size, random_state=1))

hist_df = pd.concat(hist_parts, ignore_index=True)
# HISTOGRAM_DIR = PROJECT_ROOT / "reports" / "packet_eda" / "feature_histograms"
# HISTOGRAM_DIR.mkdir(parents=True, exist_ok=True)

print("Rows used for plotting:", len(hist_df))
print("Numeric features:", len(numeric_feature_cols))
print("Categorical features:", len(categorical_feature_cols))
# print("Histogram pages are also saved to:", HISTOGRAM_DIR)
```

### Numeric feature histogram index

Each figure contains up to 12 features. Use this table to find the page containing a specific feature.

```python
numeric_page_index = []

for start in range(0, len(numeric_feature_cols), NUMERIC_PLOTS_PER_PAGE):
    page_features = numeric_feature_cols[start:start + NUMERIC_PLOTS_PER_PAGE]
    numeric_page_index.append({"page": start // NUMERIC_PLOTS_PER_PAGE + 1, "features": ", ".join(page_features)})

display(pd.DataFrame(numeric_page_index))
```

```python
def plot_numeric_histogram_pages(data, features, plots_per_page=12):
    class_order = [label for label in ["benign", "attack"] if label in data[BINARY_COL].unique()]

    for start in range(0, len(features), plots_per_page):
        page = start // plots_per_page + 1
        page_features = features[start:start + plots_per_page]
        rows = int(np.ceil(len(page_features) / 3))
        fig, axes = plt.subplots(rows, 3, figsize=(15, 3.4 * rows))
        axes = np.atleast_1d(axes).ravel()

        for ax, feature in zip(axes, page_features):
            all_values = pd.to_numeric(data[feature], errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()

            if all_values.nunique() <= 1:
                message = "No valid values" if all_values.empty else f"Constant value: {all_values.iloc[0]}"
                ax.text(0.5, 0.5, message, ha="center", va="center", transform=ax.transAxes)
                ax.set_title(feature, fontsize=10)
                ax.set_xticks([])
                ax.set_yticks([])
                continue

            lower = all_values.quantile(0.01)
            upper = all_values.quantile(0.99)
            if lower == upper:
                lower, upper = all_values.min(), all_values.max()

            plotted = False
            for label in class_order:
                values = pd.to_numeric(data.loc[data[BINARY_COL] == label, feature], errors="coerce")
                values = values.replace([np.inf, -np.inf], np.nan).dropna()
                values = values[(values >= lower) & (values <= upper)]
                if len(values) > 0:
                    ax.hist(values, bins=35, density=True, alpha=0.45, label=label)
                    plotted = True

            ax.set_title(feature, fontsize=10)
            ax.set_ylabel("Density", fontsize=8)
            ax.tick_params(axis="both", labelsize=8)
            ax.grid(alpha=0.25)
            if plotted and ax is axes[0]:
                ax.legend(fontsize=8)

        for ax in axes[len(page_features):]:
            ax.axis("off")

        fig.suptitle(f"Numeric Feature Histograms — Page {page}", fontsize=15, y=1.01)
        fig.tight_layout()
        # fig.savefig(HISTOGRAM_DIR / f"numeric_histograms_page_{page:02d}.png", dpi=150, bbox_inches="tight")
        plt.show()
        plt.close(fig)

plot_numeric_histogram_pages(hist_df, numeric_feature_cols, NUMERIC_PLOTS_PER_PAGE)
```

### Categorical feature distribution index

For each common category, the stacked bar shows the percentage of benign and attack packets. The packet count is included in the category label. This helps reveal potentially useful categories, rare values, identity-like fields, and possible leakage.

```python
categorical_page_index = []

for start in range(0, len(categorical_feature_cols), CATEGORICAL_PLOTS_PER_PAGE):
    page_features = categorical_feature_cols[start:start + CATEGORICAL_PLOTS_PER_PAGE]
    categorical_page_index.append({"page": start // CATEGORICAL_PLOTS_PER_PAGE + 1, "features": ", ".join(page_features)})

display(pd.DataFrame(categorical_page_index))
```

```python
def plot_categorical_distribution_pages(data, features, plots_per_page=4, top_n=8):
    class_order = [label for label in ["benign", "attack"] if label in data[BINARY_COL].unique()]

    for start in range(0, len(features), plots_per_page):
        page = start // plots_per_page + 1
        page_features = features[start:start + plots_per_page]
        rows = int(np.ceil(len(page_features) / 2))
        fig, axes = plt.subplots(rows, 2, figsize=(16, 4.8 * rows))
        axes = np.atleast_1d(axes).ravel()

        for ax, feature in zip(axes, page_features):
            values = data[feature].fillna("missing").astype(str)
            top_values = values.value_counts().head(top_n).index
            temp = pd.DataFrame({"value": values, BINARY_COL: data[BINARY_COL]})
            table = pd.crosstab(temp["value"], temp[BINARY_COL]).reindex(top_values).fillna(0)
            table = table.reindex(columns=class_order, fill_value=0)
            counts = table.sum(axis=1).astype(int)
            percentages = table.div(counts.replace(0, np.nan), axis=0).fillna(0) * 100

            labels = [f"{str(value)[:28]} (n={counts.loc[value]:,})" for value in percentages.index]
            percentages.index = labels
            percentages.plot(kind="barh", stacked=True, ax=ax, legend=False)

            ax.set_xlim(0, 100)
            ax.set_xlabel("Class composition (%)", fontsize=8)
            ax.set_ylabel("")
            ax.set_title(feature, fontsize=10)
            ax.tick_params(axis="both", labelsize=8)
            ax.grid(axis="x", alpha=0.25)

        for ax in axes[len(page_features):]:
            ax.axis("off")

        handles, labels = axes[0].get_legend_handles_labels()
        if handles:
            fig.legend(handles, labels, loc="upper right", bbox_to_anchor=(0.99, 0.99))

        fig.suptitle(f"Categorical Feature Distributions — Page {page}", fontsize=15, y=1.01)
        fig.tight_layout()
        # fig.savefig(HISTOGRAM_DIR / f"categorical_distributions_page_{page:02d}.png", dpi=150, bbox_inches="tight")
        plt.show()
        plt.close(fig)

plot_categorical_distribution_pages(hist_df, categorical_feature_cols, CATEGORICAL_PLOTS_PER_PAGE, TOP_CATEGORIES)
```

### How to read these plots

A feature may be useful when the benign and attack distributions have noticeably different centers, spreads, peaks, or category compositions. Strong overlap does not prove that a feature is useless because it may still help when combined with other features. Identity fields and categories that almost directly reveal the traffic source should be treated cautiously because they may not generalize to new devices or captures.

```python

summary_rows = []

for label, group in df.groupby(BINARY_COL):
    for feature in eda_features:
        values = pd.to_numeric(group[feature], errors="coerce").replace([np.inf, -np.inf], np.nan)
        summary_rows.append({
            BINARY_COL: label,
            "feature": feature,
            "median": values.median(),
            "q1": values.quantile(0.25),
            "q3": values.quantile(0.75),
            "p95": values.quantile(0.95),
            "zero_percentage": (values == 0).mean() * 100
        })

class_numeric_summary = pd.DataFrame(summary_rows)
display(class_numeric_summary.round(4))
```

## Skewness and possible outliers

```python

outlier_rows = []

for feature in eda_features:
    values = pd.to_numeric(df[feature], errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outlier_rows.append({
        "feature": feature,
        "skewness": values.skew(),
        "minimum": values.min(),
        "median": values.median(),
        "p99": values.quantile(0.99),
        "maximum": values.max(),
        "iqr_outlier_percentage": ((values < lower) | (values > upper)).mean() * 100
    })

outlier_summary = pd.DataFrame(outlier_rows).sort_values("skewness", key=lambda x: x.abs(), ascending=False)
display(outlier_summary.round(4))
```


Large values are not automatically errors in network traffic. Flooding attacks, retransmissions, bursts, and long idle periods can naturally produce extreme observations. Instead of deleting all IQR outliers, safer first choices are log transformation for non-negative skewed features, robust scaling, or clipping limits learned from the training split.


## Main preprocessing implications

Based on this EDA, the initial packet-level preprocessing pipeline should follow these rules:

1. Remove `binary_label`, `attack_type`, and `source_file` from model inputs. Labels remain available only for validation and evaluation.
2. Exclude raw identity and high-cardinality text fields from the first baseline, including IP addresses, MAC addresses, device names, hosts, URIs, and user agents.
3. Replace positive and negative infinity with missing values before imputation.
4. Treat missing rolling variance and jitter values as structural when they result from insufficient packet history; zero filling is reasonable for those fields.
5. Learn all other imputations, clipping limits, transformations, and scaling parameters from the training split only.
6. Apply `log1p` to strongly right-skewed non-negative features when it improves validation performance.
7. Reduce redundant rolling-window features instead of giving several highly correlated versions of the same behavior equal weight.
8. Preserve the original class imbalance for evaluation and use security-focused metrics rather than accuracy alone.
